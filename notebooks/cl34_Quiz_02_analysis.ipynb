## How to read a documentation page

Every function or method you look up today has a page with the same parts. You already know all of them from functions you've written yourself:

| Part of the page | What it tells you |
|---|---|
| **Signature** | The name and the parameters, like the first line of a `def` |
| **Parameters** | What each argument means. A parameter with `=` after it has a default value, so you can leave it out |
| **Returns** | The type of the value you get back |
| **Examples** | Short code you can copy and change. Start here when you're stuck! |
